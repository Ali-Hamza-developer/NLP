### NLTK data download
NLTK ke tokenizer, tagger, stopwords, wordnet, NER chunker — sab alag se download karne padte hain.

````python
import nltk

# Zaroori NLTK datasets/models download kar rahe hain
nltk.download('punkt')                          # sentence & word tokenizer
nltk.download('punkt_tab')                       # newer nltk versions ke liye
nltk.download('averaged_perceptron_tagger')       # POS tagger
nltk.download('averaged_perceptron_tagger_eng')   # POS tagger (english specific, newer versions)
nltk.download('stopwords')                        # common stopwords list
nltk.download('wordnet')                          # lemmatization ke liye
nltk.download('maxent_ne_chunker')                 # Named Entity chunker
nltk.download('maxent_ne_chunker_tab')             # newer nltk versions ke liye
nltk.download('words')                             # NER ke liye word list
````

````python
import spacy

# Downloaded model load karna
nlp = spacy.load('en_core_web_sm')
print('spaCy model loaded successfully')
````

<h2>Step 2: Sample text jispe dono libraries test karenge</h2>

````python
text = "Apple is looking at buying a U.K. startup for $1 billion in September 2025. Tim Cook wasn't happy about the delay."
text
````

<h2>(1) Tokenization</h2>

**-----NLTK-----**

````python
from nltk.tokenize import sent_tokenize, word_tokenize

# Sentence tokenization: text ko sentences mein tod na
nltk_sentences = sent_tokenize(text)
nltk_sentences
````

````python
# Word tokenization: text ko words mein tod na
nltk_words = word_tokenize(text)
nltk_words
````

**-----spaCy-----**

````python
doc = nlp(text)  # ek dafa nlp() call karne se pura pipeline (tokenize + tag + ner sab) chal jata hai

# spaCy mein sentences 'doc.sents' se milte hain
spacy_sentences = [sent.text for sent in doc.sents]
spacy_sentences
````

````python
# spaCy mein tokens 'doc' object ko loop karke milte hain
spacy_words = [token.text for token in doc]
spacy_words
````

<h2>(2) Stopwords Removal</h2>

**-----NLTK-----** (khud filter karna padta hai)

````python
from nltk.corpus import stopwords

nltk_stopwords = set(stopwords.words('english'))
filtered_nltk = [w for w in nltk_words if w.lower() not in nltk_stopwords and w.isalpha()]
filtered_nltk
````

**-----spaCy-----** (har token ka `is_stop` attribute already hota hai)

````python
filtered_spacy = [token.text for token in doc if not token.is_stop and token.is_alpha]
filtered_spacy
````

<h2>(3) POS (Part-of-Speech) Tagging</h2>

**-----NLTK-----**

````python
from nltk import pos_tag

nltk_pos = pos_tag(word_tokenize("Tim Cook wasn't happy about the delay."))
nltk_pos
````

**-----spaCy-----** (readable tags: NOUN, VERB, PROPN, etc.)

````python
doc2 = nlp("Tim Cook wasn't happy about the delay.")
spacy_pos = [(token.text, token.pos_) for token in doc2]
spacy_pos
````

<h2>(4) Lemmatization</h2>

**-----NLTK-----** (POS tag manually batana padta hai warna accuracy kam hoti hai)

````python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
words_to_lemmatize = ['running', 'better', 'wolves', 'was', 'studies']

# 'v' = verb, default noun maan leta hai agar POS na diya jaye
nltk_lemmas = [lemmatizer.lemmatize(w, pos='v') for w in words_to_lemmatize]
nltk_lemmas
````

**-----spaCy-----** (context se khud samajh leta hai, extra POS dena zaroori nahi)

````python
doc3 = nlp("running better wolves was studies")
spacy_lemmas = [token.lemma_ for token in doc3]
spacy_lemmas
````

<h2>(5) Named Entity Recognition (NER)</h2>

**-----NLTK-----** (thoda manual process — chunk tree banta hai)

````python
from nltk import ne_chunk

nltk_tree = ne_chunk(pos_tag(word_tokenize(text)))

# Tree se sirf named entities nikal rahe hain
nltk_entities = [(' '.join(c[0] for c in chunk), chunk.label())
                  for chunk in nltk_tree if hasattr(chunk, 'label')]
nltk_entities
````

**-----spaCy-----** (`doc.ents` seedha ready milta hai, labels bhi zyada detailed hain)

````python
spacy_entities = [(ent.text, ent.label_) for ent in doc.ents]
spacy_entities
````

<h2>(6) Speed — quick comparison</h2>

````python
import time

long_text = text * 50  # bada text banate hain testing ke liye

# NLTK timing
start = time.time()
_ = word_tokenize(long_text)
nltk_time = time.time() - start

# spaCy timing
start = time.time()
_ = nlp(long_text)
spacy_time = time.time() - start

print(f"NLTK tokenization time: {nltk_time:.4f} sec")
print(f"spaCy full pipeline time: {spacy_time:.4f} sec (tokenize + POS + NER sab ek sath)")
````

<h2>Summary — NLTK vs spaCy</h2>

| Feature | NLTK | spaCy |
|---|---|---|
| Design | Research/teaching-focused, bohot options | Production-focused, opinionated (ek best tareeka) |
| Speed | Slower (pure Python jyada) | Fast (Cython-optimized) |
| Setup | Har cheez alag download (`nltk.download`) | Ek model download, sab kuch bundled |
| POS tags | Short codes (NN, VB, JJ) | Readable (NOUN, VERB, ADJ) |
| Lemmatization | Manual POS dena padta hai | Automatic, context-aware |
| NER | Basic, manual tree parsing | Built-in, accurate, easy to use |
| Best for | Learning NLP concepts, custom algorithms, research | Real-world apps, production pipelines |

### Kab kya use karo?
- **NLTK** — jab NLP concepts seekhne hain, custom algorithms try karne hain, ya classic text-processing tasks (stemming, chunking) karne hain
- **spaCy** — jab production app banani hai, speed chahiye, ya ready-made accurate NER/POS/dependency parsing chahiye
