<h2 align="center">Spacy Tokenization Tutorial</h2>

```python
# spaCy library import kar rahe hain
import spacy
```

Create blank language object and tokenize words in a sentence

```python
# Blank English language object banaya (sirf tokenizer milega, koi pipeline component nahi)
nlp = spacy.blank("en")

# Text ko process karke Doc object mil raha hai
doc = nlp("Dr. Strange loves pav bhaji of mumbai as it costs only 2$ per plate.")

# Doc ke har token ko print kar rahe hain
for token in doc:
    print(token)
```

<h3>Customizing tokenizer</h3>

```python
# ORTH — special case tokenization rules define karne ke liye chahiye
from spacy.symbols import ORTH

nlp = spacy.blank("en")
doc = nlp("gimme double cheese extra large healthy pizza")

# default tokenizer "gimme" ko ek hi token treat karega
tokens = [token.text for token in doc]
tokens
```

```python
# "gimme" ko manually 2 tokens ("gim" + "me") mein split karne ka special rule add kar rahe hain
nlp.tokenizer.add_special_case("gimme", [
    {ORTH: "gim"},
    {ORTH: "me"},
])

doc = nlp("gimme double cheese extra large healthy pizza")

# ab "gimme" split ho jayega
tokens = [token.text for token in doc]
tokens
```

<h3>Sentence Tokenization or Segmentation</h3>

```python
# NOTE: blank pipeline mein sentence boundaries set nahi hote, isliye ye error dega
# (neeche "sentencizer" add karke fix kiya gaya hai)
doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")
for sentence in doc.sents:
    print(sentence)
```

```python
# pipeline components dekh rahe hain (abhi tak empty)
nlp.pipeline
```

```python
# sentence boundary detect karne ke liye 'sentencizer' component pipeline mein add kar rahe hain
nlp.add_pipe('sentencizer')
```

```python
# NOTE: blank pipeline mein sentence boundaries set nahi hote, isliye ye error dega
# (neeche "sentencizer" add karke fix kiya gaya hai)
doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")
for sentence in doc.sents:
    print(sentence)
```
